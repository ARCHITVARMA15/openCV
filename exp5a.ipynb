```python
import cv2 
import matplotlib.pyplot as plt 

img = cv2.imread('images/image3.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
#Observe image properties 
import cv2 

print("Shape:" , img.shape)
print("Height:" , img.shape[0])
print("Width:", img.shape[1])
print("Channels:", img.shape[2])
print("Data Type:", img.dtype)
print("Size:", img.size)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
plt.hist(gray.ravel(),256,[0,256])
plt.show()
```

```python
import numpy as np 
```

```python
r_min = np.min(gray)
r_max = np.max(gray)
print(r_min)
print(r_max)


if r_max == r_min:
    stretched_manual = gray.copy()
else:
    stretched_manual = ((gray - r_min)*(255/(r_max -r_min))).astype(np.uint8)


cv2.imwrite("strectched_manual.jpg", stretched_manual)
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(stretched_manual, cmap="gray", vmin=0, vmax=255)
plt.title("Strecthed Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.hist(stretched_manual.ravel(),bins =256, range=(0,255), color='black')
plt.title("Histogram After Strechung")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

r_min = np.min(stretched_manual)
r_max = np.max(stretched_manual)
print(r_min)
print(r_max)
```
